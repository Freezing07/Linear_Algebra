```python
'''
1. 用 `np.random.randn(100, 3)` 创建一个 100x3 的随机矩阵 `A`。计算它的秩。
2. 取 `x = np.array([1, 2, 3])`，计算 `b = A @ x`。用列视角验证这个结果。
3. 尝试用 `np.linalg.solve(A, b)` 解方程。为什么会报错？`A` 的形状是什么？换成方阵（比如 3x3）再试一次。
'''
import numpy as np
import matplotlib.pyplot as plt

# 1. 创建一个 100x3 的随机矩阵 A
A = np.random.randn(100, 3)
# 计算它的秩
rank_A = np.linalg.matrix_rank(A)
print("A 的秩为：", rank_A)
# 2. 取 x = np.array([1, 2, 3])，计算 b = A @ x
x = np.array([1, 2, 3])
b = A @ x
# 用列视角验证这个结果
print("b 的值为：", b)

# 3. 尝试用 np.linalg.solve(A, b) 解方程
try:
    solution = np.linalg.solve(A, b)
    print("解为：", solution)
except np.linalg.LinAlgError:
    print("A 不是方阵，无法求解")

# 换成方阵（比如 3x3）再试一次
A = np.random.randn(3, 3)
print(f"A 的形状为：{A.shape}")
b = A @ x
solution = np.linalg.solve(A, b)
print("解为：", solution)
```
