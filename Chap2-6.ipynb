```python
'''1. 用 `np.random.randn(5, 5)` 创建一个 5x5 的随机矩阵 A，用 `np.random.randn(5)` 创建 b。用 `solve` 解方程。再用手算验证一下前向替换和回代的过程。
2. 创建一个 4x2 的矩阵 A（4行2列），和 4维向量 b。用 `lstsq` 解。计算残差 `||Ax - b||`。这个残差是什么几何意义？
3. 用 LU 分解解同一个方程，验证结果一致。'''

import numpy as np
from scipy.linalg import solve, lstsq, lu

# 1
A = np.random.randn(5, 5)
b = np.random.randn(5)
x = solve(A, b)
print(f"x = {x}")

# 手算前向替换和回代
P, L, U = lu(A)
# LU = Pb
y = np.dot(P, b)
z = np.linalg.solve(L, y)
x = np.linalg.solve(U, z)

print(f"x = {x}")

# 2
A = np.random.randn(4, 2)
b = np.random.randn(4)
x, residuals, rank, s = lstsq(A, b)
print(f"x = {x}")
print(f"residuals = {residuals}")
print(f"rank = {rank}")
print(f"s = {s}")

# 3
A = np.random.randn(5, 5)
b = np.random.randn(5)
P, L, U = lu(A)
x = np.linalg.solve(U, np.linalg.solve(L, np.dot(P, b)))
print(f"\n分解是否正确? {np.allclose(A, P @ L @ U)}")
```
