```python
'''1. 创建一组正交向量（用 Gram-Schmidt），验证它们两两正交。
2. 用投影矩阵 `P = A (A^T A)^{-1} A^T` 投影一个向量到列空间上。
3. 用 QR 分解（`np.linalg.qr`）解一个最小二乘问题，验证结果与 `np.linalg.lstsq` 一致。
'''

import numpy as np

# 1. 创建一组正交向量（用 Gram-Schmidt），验证它们两两正交。
A = np.random.randn(3, 3)
print('rank(A):\n', np.linalg.matrix_rank(A))
Q, R = np.linalg.qr(A)
print('Q:\n', Q)
print(f'q1 cdot q2 = {np.dot(Q[:, 0], Q[:, 1])}')
print(f'q1 cdot q3 = {np.dot(Q[:, 0], Q[:, 2])}')
print(f'q2 cdot q3 = {np.dot(Q[:, 1], Q[:, 2])}')

# 2. 用投影矩阵 `P = A (A^T A)^{-1} A^T` 投影一个向量到列空间上。
b = np.random.randn(3)
P = A @ np.linalg.inv(A.T @ A) @ A.T
print('P:\n', P)
print('P @ b:\n', P @ b)

# 3. 用 QR 分解（`np.linalg.qr`）解一个最小二乘问题，验证结果与 `np.linalg.lstsq` 一致。
print('='*50)
print('Use np.linalg.lstsq')
print('='*50)
b = np.random.randn(3)
x_lstsq, residuals_lstsq, rank_lstsq, s_lstsq = np.linalg.lstsq(A, b, rcond=None)
print('x:\n', x_lstsq)
print('residuals:\n', residuals_lstsq)
print('rank:\n', rank_lstsq)
print('s:\n', s_lstsq)

# QR 分解：QRx = b,Rx = Q^Tb
print('='*50)
print('Use QR decomposition')
print('='*50)
Q, R = np.linalg.qr(A)
x_qr = np.linalg.solve(R, Q.T @ b)
p_qr = A @ x_qr
e_qr = b - p_qr
print('x:\n', x_qr)
print('Project(p) :\n', p_qr)
print('Residual(e) :\n', e_qr)
print('Residual norm :\n', np.linalg.norm(e_qr, ord=2))
print('Rank :\n', np.linalg.matrix_rank(Q))
```
