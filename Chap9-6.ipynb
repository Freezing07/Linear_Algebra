```python
'''1. 推导线性回归的损失函数是二次函数，写出对应的 $A$ 和 $b$。
2. 实现梯度下降（不调用 `sklearn`），用你生成的数据测试。
3. 比较普通最小二乘和岭回归在共线性数据下的表现。
4. 用拉格朗日乘子法求解 $x_1^2 + 2x_2^2$ 在 $x_1 + x_2 = 1$ 约束下的最优解。
'''

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error
from scipy.optimize import minimize

# 1. 推导线性回归的损失函数是二次函数，写出对应的 A 和 b。
# Cost Function: ||e||^2 = (y - Xw)^T(y - Xw)
# Gradient: d||e||^2/dw = 2X^T(Xw - y)
# A = 2X^TX, b = 2X^Ty

# 2. 实现梯度下降（不调用 sklearn），用你生成的数据测试。
# Generate data
X = np.array([[1, 1], [1, 2], [1, 3], [1, 4]])
y = np.dot(X, np.array([1, 2])) + np.random.randn(4) * 0.1

# Initialize parameters
w = np.zeros(2)

# Gradient descent
learning_rate = 0.01
num_iterations = 1000
for i in range(num_iterations):
    gradient = 2 * np.dot(X.T, np.dot(X, w) - y)
    w -= learning_rate * gradient

# Print the final parameters
print("Final parameters:", w)

# 3. 比较普通最小二乘和岭回归在共线性数据下的表现。
# Generate data with multicollinearity
np.random.seed(42)
X_multicollinearity = np.array([[1, 1], [1, 2], [1, 3], [1, 4]])
y_multicollinearity = np.dot(X_multicollinearity, np.array([1, 2])) + np.random.randn(4) * 0.1

# Fit models
model_ordinary = LinearRegression().fit(X_multicollinearity, y_multicollinearity)
model_ridge = Ridge(alpha=0).fit(X_multicollinearity, y_multicollinearity) # \lambda=0

# Print the coefficients
print("Ordinary least squares coefficients:", model_ordinary.coef_)
print("Ridge regression coefficients:", model_ridge.coef_)

# 4. 拉格朗日乘子法求解 $x_1^2 + 2x_2^2$ 在 $x_1 + x_2 = 1$ 约束下的最优解。
# f(x) = x_1^2 + 2x_2^2 -> [x_1, x_2] * [[1, 0],[0, 2]] * [x_1, x_2] = x^T * A * x
# Constraint g(x): x_1 + x_2 = 1 -> [1, 1] * [x_1, x_2] = 1 -> C * x - d = 0
# Lagrange multiplier method
# L(x, λ) = f(x) + λ * g(x) = x^T * A * x + λ^T * (C * x - d)
# ∂L/∂x = 2 * A * x + λ * C^T = 0
# ∂L/∂λ = C * x - d = 0
# Solve for x and λ
# x = (- λ/2) * A^(-1) * C^T -> (- λ/2) * C * A^(-1) * C^T = d
# λ = -2 * (C * A^(-1) * C^T)^(-1) * d
# x = -(λ/2) * A^(-1) * C^T = A^(-1) * C^T (C * A^(-1) * C^T)^(-1) * d
A = np.array([[1, 0], [0, 2]])
C = np.array([[1, 1]])
d = np.array([[1]])
x = np.linalg.inv(A) @ C.T @ np.linalg.inv(C @ np.linalg.inv(A) @ C.T) @ d
print("Lagrange multiplier method solution:\n", x)
```
