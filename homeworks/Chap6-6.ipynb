```python
'''1. 创建几个不同的 2x2 矩阵，计算它们的特征值和特征向量。在纸上画出特征向量的方向。
2. 创建一个对称矩阵，验证它的特征向量正交。
3. 创建一个非正定的矩阵，计算它的特征值，判断它是什么“形状”（鞍点？负定？）。
4. 用特征值分解计算 $A^10$，与 `np.linalg.matrix_power` 的结果对比。'''

import numpy as np
import matplotlib.pyplot as plt

# 1. 创建几个不同的 2x2 矩阵，计算它们的特征值和特征向量。在纸上画出特征向量的方向。
A1 = np.array([[1, 2], [3, 4]])
A2 = np.array([[5, 6], [7, 8]])

# 计算特征值和特征向量
eigenvalues1, eigenvectors1 = np.linalg.eig(A1)
eigenvalues2, eigenvectors2 = np.linalg.eig(A2)

# 在纸上画出特征向量的方向
plt.figure(figsize=(10, 5))

# 绘制特征向量1
plt.subplot(1, 2, 1)
plt.quiver(0, 0, eigenvectors1[0, 0], eigenvectors1[1, 0], color='r', label='Eigenvector 1')
plt.quiver(0, 0, eigenvectors1[0, 1], eigenvectors1[1, 1], color='b', label='Eigenvector 2')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Eigenvectors of A1')
plt.legend()

# 绘制特征向量2
plt.subplot(1, 2, 2)
plt.quiver(0, 0, eigenvectors2[0, 0], eigenvectors2[1, 0], color='r', label='Eigenvector 1')
plt.quiver(0, 0, eigenvectors2[0, 1], eigenvectors2[1, 1], color='b', label='Eigenvector 2')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Eigenvectors of A2')
plt.legend()

plt.show()

# 2. 创建一个对称矩阵，验证它的特征向量正交。
A = np.array([[1, 2], [2, 3]])

# 计算特征值和特征向量
eigenvalues, eigenvectors = np.linalg.eigh(A)

# 验证特征向量正交
dot_product = np.dot(eigenvectors[:, 0], eigenvectors[:, 1])
print("Dot product of eigenvectors:", dot_product)

# 3. 创建一个非正定的矩阵，计算它的特征值，判断它是什么“形状”（鞍点？负定？）。
A = np.array([[1, -2], [-2, 1]])

# 计算特征值
eigenvalues = np.linalg.eigvals(A)
print("Eigenvalues of A:", eigenvalues)

# 判断矩阵的形状
if np.all(eigenvalues < 0):
    print("The matrix is negative definite.")
elif np.all(eigenvalues > 0):
    print("The matrix is positive definite.")
else:
    print("The matrix is indefinite, possibly containing a saddle point.")

# 4. 用特征值分解计算 $A^10$，与 `np.linalg.matrix_power` 的结果对比。
A = np.array([[1, 2], [3, 4]])

# 计算特征值和特征向量
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:", eigenvectors)

# 计算特征向量的幂：X \Lamda^10 X^{-1}
eigenvectors_power = (eigenvectors * (eigenvalues ** 10)) @ np.linalg.inv(eigenvectors)

# 计算A的10次幂
A_power = np.linalg.matrix_power(A, 10)

print(f"A^10 calculated using eigenvalue decomposition:\n{eigenvectors_power}")
print(f"A^10 calculated using np.linalg.matrix_power:\n{A_power}")
```
