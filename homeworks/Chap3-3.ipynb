```python
'''
1. 用 `np.random.randn(10, 5)` 创建一个 10×5 的随机矩阵。计算它的秩。
2. 把第 5 列设为前 4 列的线性组合（比如 `col5 = 2*col1 - col3 + 0.5*col4`）。重新计算秩。发生了什么？
3. 用 `null_space` 找出零空间的基。这个基告诉你第5列和其他列的关系是什么。
4. 用这个矩阵做一个线性回归（用 `np.linalg.lstsq`）。然后加上一个零空间向量到系数上，验证预测值不变。'''

import numpy as np
from scipy.linalg import null_space

# 1. 用 `np.random.randn(10, 5)` 创建一个 10×5 的随机矩阵。计算它的秩。
A = np.random.randn(10, 5)
# print(A)
print(f'rank of A is {np.linalg.matrix_rank(A)}')

# 2. 把第 5 列设为前 4 列的线性组合（比如 `col5 = 2*col1 - col3 + 0.5*col4`）。重新计算秩。发生了什么？
A[:, 4] = 2*A[:, 0] - A[:, 2] + 0.5*A[:, 3]
print(f'rank of A is {np.linalg.matrix_rank(A)}')

# 3. 用 `null_space` 找出零空间的基。这个基告诉你第5列和其他列的关系是什么。
print(f'null space of A is {null_space(A)}')

# 4. 用这个矩阵做一个线性回归（用 `np.linalg.lstsq`）。然后加上一个零空间向量到系数上，验证预测值不变。
# np.random.seed(42)
X = A
y = np.random.randn(10)
coefficients, residuals, rank, s = np.linalg.lstsq(X, y, rcond=None)
print(f'coefficients: {coefficients}')
print(f'rank: {rank}')
print(f'residuals: {residuals}')
print(f's: {s}')
original_predictions = np.dot(X, coefficients)
null_space_vector = null_space(A)[:, 0]
new_coefficients = coefficients + null_space_vector
print(f'new coefficients: {new_coefficients}')
new_predictions = np.dot(X, new_coefficients)
print(f'new predictions: {new_predictions}')
print(f'original predictions: {original_predictions}')
print(f'are they equal? {np.allclose(new_predictions, original_predictions)}')
```
