```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
x = np.array([5, 6])

print("A = \n", A)
print("x =", x)

# ---- 视角1：行视角（点积）----
print("\n=== 视角1：行视角（点积）===")
b_row = np.array([
    np.dot(A[0, :], x),  # 第一行和 x 的点积
    np.dot(A[1, :], x)   # 第二行和 x 的点积
])
print(f"行视角结果: {b_row}")
print(f"A @ x 结果: {A @ x}")

# ---- 视角2：列视角（线性组合）----
print("\n=== 视角2：列视角（线性组合）===")
# b = x1 * (列1) + x2 * (列2)
col1 = A[:, 0]  # [1, 3]
col2 = A[:, 1]  # [2, 4]
b_col = x[0] * col1 + x[1] * col2
print(f"列视角结果: {b_col}")
print(f"即: {x[0]} * {col1} + {x[1]} * {col2} = {b_col}")

# ---- 视角3：行变换视角（对 A 的行做组合）----
# 这本质上与列视角相同，但换了个角度
# 见实操中的说明

# ---- 视角4：列-行外积（用于矩阵乘法 AB）----
print("\n=== 视角4：列-行外积（矩阵乘法）===")
B = np.array([[7, 8], [9, 10]])
print("B = \n", B)

# AB = (A的列1 × B的行1) + (A的列2 × B的行2)
outer1 = np.outer(A[:, 0], B[0, :])  # 列1 × 行1
outer2 = np.outer(A[:, 1], B[1, :])  # 列2 × 行2
AB_outer = outer1 + outer2
print(f"外积视角计算 AB:\n{AB_outer}")
print(f"直接用 A @ B:\n{A @ B}")
```

```python
# 单元格1：导入库和通用配置
from manim import *
import numpy as np

# 配置Manim在Jupyter中渲染
config.media_width = "100%"
config.quality = "high_quality"

class MatrixMultiplicationBase(Scene):
    """基类，提供通用配置"""
    def setup(self):
        # 定义矩阵A和向量x
        self.A = np.array([[1, 2], [3, 4]])
        self.x = np.array([5, 6])
        # 颜色方案
        self.colors = [BLUE, GREEN, RED, PURPLE]
```

```python
# 单元格2：视角1 - 行视角（点积）
class RowPerspective(MatrixMultiplicationBase):
    def construct(self):
        # 标题
        title = Text("视角1：行视角（点积）", font_size=36, color=YELLOW)
        title.to_edge(UP)
        self.play(Write(title))
        self.wait(0.5)
        
        # 显示矩阵A和向量x
        a_matrix = Matrix(self.A, v_buff=0.8, h_buff=1.0)
        a_matrix.next_to(title, DOWN, buff=0.5)
        a_matrix.shift(LEFT * 2.5)
        a_label = Tex("A =", font_size=30).next_to(a_matrix, LEFT)
        
        x_vector = Matrix([self.x], v_buff=0.8, h_buff=1.0)
        x_vector.next_to(a_matrix, RIGHT, buff=1.0)
        x_label = Tex("x =", font_size=30).next_to(x_vector, LEFT)
        
        self.play(
            Write(a_label),
            Write(a_matrix),
            Write(x_label),
            Write(x_vector)
        )
        self.wait(0.5)
        
        # 逐行演示点积
        rows = [row for row in self.A]
        results = [np.dot(row, self.x) for row in rows]
        
        # 高亮第一行
        row1_highlight = SurroundingRectangle(
            a_matrix.get_entries()[0:2],
            color=BLUE,
            buff=0.15
        )
        self.play(Create(row1_highlight))
        self.wait(0.3)
        
        # 显示第一行的计算过程
        calc1 = MathTex(
            f"{self.A[0,0]} \\times {self.x[0]} + {self.A[0,1]} \\times {self.x[1]} = {results[0]}",
            color=BLUE
        )
        calc1.next_to(a_matrix, DOWN, buff=0.8)
        self.play(Write(calc1))
        self.wait(0.5)
        
        # 高亮第二行
        row2_highlight = SurroundingRectangle(
            a_matrix.get_entries()[2:4],
            color=GREEN,
            buff=0.15
        )
        self.play(
            FadeOut(row1_highlight),
            Create(row2_highlight)
        )
        self.wait(0.3)
        
        # 显示第二行的计算过程
        calc2 = MathTex(
            f"{self.A[1,0]} \\times {self.x[0]} + {self.A[1,1]} \\times {self.x[1]} = {results[1]}",
            color=GREEN
        )
        calc2.next_to(calc1, DOWN, buff=0.3)
        self.play(Write(calc2))
        self.wait(0.5)
        
        # 显示最终结果
        result = Matrix([results], v_buff=0.8, h_buff=1.0)
        result.next_to(x_vector, RIGHT, buff=1.0)
        result_label = Tex("b =", font_size=30).next_to(result, LEFT)
        
        result_box = SurroundingRectangle(result, color=YELLOW, buff=0.3)
        
        self.play(
            FadeOut(row2_highlight),
            Write(result_label),
            Write(result),
            Create(result_box)
        )
        self.wait(0.5)
        
        # 添加说明
        note = Text(
            "每一行与x做点积 → 得到b的对应元素",
            font_size=24,
            color=WHITE
        )
        note.next_to(result, DOWN, buff=0.8)
        self.play(Write(note))
        self.wait(2)

# 在Jupyter中渲染
RowPerspective().render()
```

```python
# 单元格3：视角2 - 列视角（线性组合）- 最重要的视角
class ColumnPerspective(MatrixMultiplicationBase):
    def construct(self):
        # 标题
        title = Text("视角2：列视角（线性组合）", font_size=36, color=YELLOW)
        title.to_edge(UP)
        self.play(Write(title))
        self.wait(0.5)
        
        # 显示矩阵A和向量x
        a_matrix = Matrix(self.A, v_buff=0.8, h_buff=1.0)
        a_matrix.next_to(title, DOWN, buff=0.5)
        a_matrix.shift(LEFT * 2.5)
        a_label = Tex("A =", font_size=30).next_to(a_matrix, LEFT)
        
        x_vector = Matrix([self.x], v_buff=0.8, h_buff=1.0)
        x_vector.next_to(a_matrix, RIGHT, buff=1.0)
        x_label = Tex("x =", font_size=30).next_to(x_vector, LEFT)
        
        self.play(
            Write(a_label),
            Write(a_matrix),
            Write(x_label),
            Write(x_vector)
        )
        self.wait(0.5)
        
        # 提取列并高亮
        col1 = self.A[:, 0]
        col2 = self.A[:, 1]
        
        # 高亮第一列
        col1_entries = [
            a_matrix.get_entries()[0],
            a_matrix.get_entries()[2]
        ]
        col1_highlight = VGroup(*[
            SurroundingRectangle(entry, color=BLUE, buff=0.1)
            for entry in col1_entries
        ])
        col1_label = Tex("Column 1", color=BLUE, font_size=24).next_to(
            a_matrix, DOWN, buff=0.2
        ).shift(LEFT * 0.5)
        
        self.play(
            Create(col1_highlight),
            Write(col1_label)
        )
        self.wait(0.5)
        
        # 显示列1的线性组合
        combo1 = MathTex(
            f"{self.x[0]} \\times \\begin{{bmatrix}} {col1[0]} \\\\ {col1[1]} \\end{{bmatrix}}",
            color=BLUE
        )
        combo1.next_to(a_matrix, DOWN, buff=1.0)
        self.play(Write(combo1))
        self.wait(0.3)
        
        # 高亮第二列
        col2_entries = [
            a_matrix.get_entries()[1],
            a_matrix.get_entries()[3]
        ]
        col2_highlight = VGroup(*[
            SurroundingRectangle(entry, color=GREEN, buff=0.1)
            for entry in col2_entries
        ])
        col2_label = Tex("Column 2", color=GREEN, font_size=24).next_to(
            a_matrix, DOWN, buff=0.2
        ).shift(RIGHT * 0.5)
        
        self.play(
            FadeOut(col1_highlight),
            FadeOut(col1_label),
            Create(col2_highlight),
            Write(col2_label)
        )
        self.wait(0.5)
        
        # 显示完整的线性组合
        combo2 = MathTex(
            f"+ {self.x[1]} \\times \\begin{{bmatrix}} {col2[0]} \\\\ {col2[1]} \\end{{bmatrix}}",
            color=GREEN
        )
        combo2.next_to(combo1, RIGHT)
        self.play(Write(combo2))
        self.wait(0.5)
        
        # 显示计算结果
        result = self.x[0] * col1 + self.x[1] * col2
        result_tex = MathTex(
            f"= \\begin{{bmatrix}} {result[0]} \\\\ {result[1]} \\end{{bmatrix}}",
            color=YELLOW
        )
        result_tex.next_to(combo2, RIGHT, buff=0.3)
        self.play(Write(result_tex))
        self.wait(0.5)
        
        # 添加重要说明
        note1 = Text(
            "b = x₁ × (列1) + x₂ × (列2)",
            font_size=28,
            color=YELLOW
        )
        note1.next_to(title, DOWN, buff=0.2)
        note1.shift(RIGHT * 3)
        self.play(Write(note1))
        self.wait(0.5)
        
        note2 = Text(
            "⭐ 最重要的视角：用x的系数组合A的列！",
            font_size=26,
            color=PURPLE
        )
        note2.next_to(note1, DOWN, buff=0.2)
        self.play(Write(note2))
        self.wait(2)

ColumnPerspective().render()
```

```python
%%manim -v WARNING --disable_caching -qm OuterProductPerspective

# 单元格5：视角4 - 列-行外积视角（矩阵乘法）
class OuterProductPerspective(MatrixMultiplicationBase):
    def construct(self):
        # 标题
        title = Text("视角4：列-行外积视角（矩阵乘法）", font_size=36, color=YELLOW)
        title.to_edge(UP)
        self.play(Write(title))
        self.wait(0.5)
        
        # 定义矩阵B
        B = np.array([[7, 8], [9, 10]])
        
        # 显示矩阵A和B
        a_matrix = Matrix(self.A, v_buff=0.8, h_buff=1.0)
        a_matrix.next_to(title, DOWN, buff=0.4)
        a_matrix.shift(LEFT * 3)
        a_label = Tex("A =", font_size=30).next_to(a_matrix, LEFT)
        
        b_matrix = Matrix(B, v_buff=0.8, h_buff=1.0)
        b_matrix.next_to(a_matrix, RIGHT, buff=1.5)
        b_label = Tex("B =", font_size=30).next_to(b_matrix, LEFT)
        
        self.play(
            Write(a_label),
            Write(a_matrix),
            Write(b_label),
            Write(b_matrix)
        )
        self.wait(0.5)
        
        # 提取列和行
        col1 = self.A[:, 0]  # [1, 3]
        col2 = self.A[:, 1]  # [2, 4]
        row1 = B[0, :]       # [7, 8]
        row2 = B[1, :]       # [9, 10]
        
        # 计算外积1
        outer1 = np.outer(col1, row1)
        outer2 = np.outer(col2, row2)
        result = outer1 + outer2
        
        # 创建外积1的矩阵
        outer1_matrix = Matrix(outer1, v_buff=0.6, h_buff=0.8)
        outer1_matrix.next_to(a_matrix, DOWN, buff=1.0)
        outer1_matrix.shift(LEFT * 1.5)
        outer1_label = Tex(
            f"Column 1 × Row 1 =", 
            font_size=24,
            color=BLUE
        ).next_to(outer1_matrix, LEFT)
        
        # 高亮A的第1列和B的第1行
        col1_highlight = SurroundingRectangle(
            VGroup(a_matrix.get_entries()[0], a_matrix.get_entries()[2]),
            color=BLUE,
            buff=0.1
        )
        row1_highlight = SurroundingRectangle(
            b_matrix.get_entries()[0:2],
            color=BLUE,
            buff=0.1
        )
        
        self.play(
            Create(col1_highlight),
            Create(row1_highlight)
        )
        self.wait(0.3)
        
        # 显示外积1
        self.play(
            Write(outer1_label),
            Write(outer1_matrix)
        )
        self.wait(0.3)
        
        # 创建外积2的矩阵
        outer2_matrix = Matrix(outer2, v_buff=0.6, h_buff=0.8)
        outer2_matrix.next_to(outer1_matrix, RIGHT, buff=1.2)
        outer2_label = Tex(
            f"+ Column 2 × Row 2 =", 
            font_size=24,
            color=GREEN
        ).next_to(outer2_matrix, LEFT)
        
        # 高亮A的第2列和B的第2行
        col2_highlight = SurroundingRectangle(
            VGroup(a_matrix.get_entries()[1], a_matrix.get_entries()[3]),
            color=GREEN,
            buff=0.1
        )
        row2_highlight = SurroundingRectangle(
            b_matrix.get_entries()[2:4],
            color=GREEN,
            buff=0.1
        )
        
        self.play(
            FadeOut(col1_highlight),
            FadeOut(row1_highlight),
            Create(col2_highlight),
            Create(row2_highlight)
        )
        self.wait(0.3)
        
        # 显示外积2
        self.play(
            Write(outer2_label),
            Write(outer2_matrix)
        )
        self.wait(0.3)
        
        # 显示最终结果
        result_matrix = Matrix(result, v_buff=0.6, h_buff=0.8)
        result_matrix.next_to(outer2_matrix, RIGHT, buff=1.0)
        result_label = Tex("=", font_size=40).next_to(outer2_matrix, RIGHT, buff=0.3)
        result_box = SurroundingRectangle(result_matrix, color=YELLOW, buff=0.3)
        
        self.play(
            Write(result_label),
            Write(result_matrix),
            Create(result_box)
        )
        self.wait(0.5)
        
        # 添加说明
        note1 = Text(
            "AB = 列1×行1 + 列2×行2 + ...",
            font_size=26,
            color=WHITE
        )
        note1.next_to(outer1_matrix, DOWN, buff=0.6)
        self.play(Write(note1))
        self.wait(0.3)
        
        note2 = Text(
            "这是理解SVD、秩1分解的基础！",
            font_size=24,
            color=PURPLE
        )
        note2.next_to(note1, DOWN, buff=0.2)
        self.play(Write(note2))
        self.wait(2)
```
